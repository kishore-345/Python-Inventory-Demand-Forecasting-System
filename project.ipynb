```python
import numpy as np
import tkinter as tk
from tkinter import messagebox


class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock


class DemandForecaster:
    def __init__(self, sales_data):
        self.sales_data = np.array(sales_data)

    def average_demand(self):
        return np.mean(self.sales_data)

    def demand_volatility(self):
        return np.std(self.sales_data)

    def forecast_next_month(self):
        if len(self.sales_data) >= 3:
            return np.mean(self.sales_data[-3:])
        else:
            return np.mean(self.sales_data)


def calculate_forecast():
    try:
        name = entry_name.get()
        price = float(entry_price.get())
        stock = int(entry_stock.get())

        sales_input = entry_sales.get()
        sales_list = list(map(int, sales_input.split(",")))

        product = Product(name, price, stock)
        forecaster = DemandForecaster(sales_list)

        avg = forecaster.average_demand()
        volatility = forecaster.demand_volatility()
        forecast = forecaster.forecast_next_month()

        result_text = f"""
Product: {product.name}
Price: {product.price}
Stock: {product.stock}

Average Demand: {round(avg,2)}
Demand Volatility: {round(volatility,2)}
Forecast Next Month: {round(forecast,2)}
"""

        if forecast > stock:
            result_text += f"\nReorder Required! Suggested Order: {int(forecast-stock)}"
        else:
            result_text += "\nStock is sufficient."

        result_label.config(text=result_text)

    except:
        messagebox.showerror("Error", "Please enter valid data")


# Main window
root = tk.Tk()
root.title("Inventory Forecasting System")
root.geometry("400x450")


# Bold title inside the page
title_label = tk.Label(root,
                       text="Inventory Forecasting System",
                       font=("Arial", 16, "bold"))
title_label.pack(pady=10)


tk.Label(root, text="Product Name").pack()
entry_name = tk.Entry(root)
entry_name.pack()

tk.Label(root, text="Product Price").pack()
entry_price = tk.Entry(root)
entry_price.pack()

tk.Label(root, text="Current Stock").pack()
entry_stock = tk.Entry(root)
entry_stock.pack()

tk.Label(root, text="Monthly Sales (comma separated)").pack()
entry_sales = tk.Entry(root)
entry_sales.pack()

tk.Button(root, text="Analyze Demand",
          command=calculate_forecast).pack(pady=10)

result_label = tk.Label(root, text="", justify="left")
result_label.pack()
root.mainloop()
```
